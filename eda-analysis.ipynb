### Question 4: Do senior roles have significantly higher salaries?

````python
df.groupby("level")["salary_min"].mean()
````

````python
df.groupby("level")["salary_min"].mean().plot(kind="bar")
````

### Question 6: What is the average salary offered by each company?

### Question 7: Based on the dataset, which skills are most important in hiring?

````python
skills=df[["skill_1","skill_2","skill_3","skill_4","skill_5","skill_6","skill_7","skill_8","skill_9","skill_10"]].stack()
skills.value_counts().head(3)
````

Answer: Based on how often they appear in the dataset the most important skills in hiring are Python, machine learning and ml

### Question 8: Which skills are associated with the highest salaries?

````python
skill_salary=df[["skill_1","skill_2","skill_3","skill_4","skill_5","skill_6","skill_7","skill_8","skill_9","skill_10"]].stack()
skill_salary=pd.DataFrame({"skill":skill_salary.values,"salary":df["salary_min"].repeat(10).values})
````

````python
skill_salary.groupby("skill")["salary"].mean().sort_values(ascending=False).head(3)
````

````python
skill_salary.groupby("skill")["salary"].mean().sort_values(ascending=False).head(3).plot(kind="bar")
````

Answer: Based on the dataset's minimum salaries the top 3 skills associated with the highest average pay are Kubeflow, Git and Keras.

### Question 11: What is the salary spread within each experience level?

````python
df.groupby("level")["salary_min"].std()
````

````python
df.boxplot(column="salary_min", by="level")
````

Answer: The boxplot lets us see how salaries are spread and spot the outliers for each level. The results show that junior roles have the smallest pay range, mid and senior have wider pay range with a few high pay outliers.

### Question 12: How do remote-friendly companies compare in terms of salaries?

````python
df.groupby(df['location'].str.contains('Remote',case=False,na=False))['salary_min'].mean()
````

````python
df.groupby(df['location'].str.contains('Remote',case=False,na=False))['salary_min'].mean().plot(kind='bar')
````

Answer:The bar chart compares the average starting salary for non-remote(false) and remote(true) roles. It shows that non-remote positions offer higher avarage salaries compared to remote jobs.
